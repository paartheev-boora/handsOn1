# Python Basics and data types

##### Numeric: int, float, complex
##### Sequence Type: string, list, tuple
##### Mapping Type: dict
##### Boolean: bool
##### Set Type: set, frozenset
##### Binary Types: bytes, bytearray, memoryview

```python
##### Program to check if a given number is prime.
##### I/P: Integer ; O/P: True/False
def isPrime(x):
    if x<2:
        return False
    for i in range(2, int(x/2)+1):
        if x%i==0:
            return False
    return True
        
print(isPrime(4))
print(isPrime(2))
print(isPrime(1))
```

### List Manipulation
##### Adding : append, insert, extend, +
##### removing : remove, pop, clear, del
##### searching : index, count, in
##### sorting : sort, sorted(returns a new list)
##### copying : slicing([:]), copy
##### built in : len, sum, min, max
##### advanced : filter, map, zip

```python
def even_num_of(lst):
    eve_num=[num for num in lst if num%2==0]
    eve_num.sort(reverse=True)
    return eve_num

list1=[12,113,15,17,16,19]
print(even_num_of(list1))
```

### Tuple and Set operations
##### built in : all(tup) : true if all ele are true
##### any(tup) : True if any ele are true
##### sorted(tup/set) : returns a sorted list

```python
lst=["Partheev","Geethika","Dharani","Aryan","Aryan"]
tup=tuple(lst)

unique_sparta=set(tup)

print(sorted(unique_sparta))
```

### Dictionaries

```python
dicti={name:None for name in lst}
vowels="aeiou"
dicti={name: sum(1 for ch in name if ch.lower() in vowels) for name in lst}
print(dicti)
```

### Numpy
#### np.arange(), np.array(), np.diag(), np.sum()

```python
import numpy as np
arr=np.arange(1,26).reshape(5,5)
mat=np.matrix(arr)
print(mat)
diag_ele=mat.diagonal()
print(diag_ele)
print(f"sum of diagonal ele : {np.sum(diag_ele)}")
```

### Pandas Dataframe
##### we can also use df.quantile(0.75) to calculate 75% percentile

```python
import pandas as pd
employees=pd.DataFrame({
    'name':['Geethika', 'Dharani', 'Aryan', 'harsh'],
    'dept':['associate', 'associate', 'Trainer', 'Trainer'],
    'sal':[30000, 32000, 100000, 90000]
})

summary= employees['sal'].describe()
print(summary)

perc_75=summary['75%']
print(perc_75)

emp_75=employees[employees['sal']>perc_75]
print(emp_75)
```

```python
import pandas as pd
sale=pd.DataFrame({
    'prod':['Laptop','camera','phone','headphones','speaker'],
    'category':['electronics','electronics','accessories','accessories','electronics'],
    'sales':[290000,500000,340000,120000,400000]
        })
print(sale)


sale_cat=sale.groupby('category').agg(tot_sales=('sales','sum'))
print(sale_cat)
```

```python
import matplotlib.pyplot as plt
sale_cat.plot(kind='bar')
plt.title("Total sales by product category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.legend()
plt.show()
```

### OOP- Class and inheritance

```python
class Vehicle:
    def __init__(self, make, model):
        self.make=make
        self.model=model
class Car(Vehicle):
    def __init__(self, make, model, num_doors):
        super().__init__(make, model)
        self.num_doors=num_doors
    def display(self):
        print("Make:",self.make)
        print("Model:",self.model)
        print("num_doors:",self.num_doors)

c=Car("Pagani", "Zonda R", 2)
c.display()
```

### OOP-Encapsulation and methods
##### Encapsulation is all about protecting data that should not get accessed/modified by others.
##### We can decalre private data members as __balance("using double under score")

```python
class bankAcc:
    def __init__(self, account_holder,balance=0.0):
        self.account_holder=account_holder
        self.__balance=balance #private Attribute

    def deposit(self, amount):
        if amount>0:
            self.__balance+=amount
            return True
        return False
    
    def withdrawl(self, amount):
        if 0<amount<=self.__balance:
            self.__balance-=amount
            return True
        return False
    
    def get_balance(self):
        return self.__balance
    
    def __str__(self):
        return f"Account: {self.account_holder}, Balance:{self.__balance}"
    
account1=bankAcc("Nikhil",20000)
print(account1.deposit(1000))
print(account1.get_balance())
print(account1.withdrawl(5000))
print(account1.__str__())
```
